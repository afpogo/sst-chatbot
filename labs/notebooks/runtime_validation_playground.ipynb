```python
import subprocess
import sys

result = subprocess.run(
    [sys.executable, "init.py"],
    capture_output=True,
text=True,
    cwd=".",
)

print("Return code:", result.returncode)
if result.stdout:
    print("STDOUT:\n" + result.stdout)
if result.stderr:
    print("STDERR:\n" + result.stderr)

if result.returncode != 0:
    raise RuntimeError("init.py fallo. Revisar la salida anterior.")
```

```python
import os

from dotenv import load_dotenv


def load_jupyter_settings() -> None:
    try:
        from jupyter_settings import SETTINGS
    except ImportError:
        return

    for key, value in SETTINGS.items():
        if value:
            os.environ[key] = value


load_dotenv()
load_jupyter_settings()

import openai  # noqa: F401
from langchain_openai import OpenAI  # noqa: F401

required_keys = ("OPENAI_API_KEY", "LANGCHAIN_API_KEY")
missing_keys = [key for key in required_keys if not os.getenv(key)]

if missing_keys:
    raise RuntimeError(
        "Faltan variables requeridas: "
        + ", ".join(missing_keys)
        + ". Definilas en .env o en jupyter_settings.py."
    )

print("Configuracion completada y validada con exito.")
```
